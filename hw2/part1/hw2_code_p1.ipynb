# Evaluation code

In the test phase, you will be given 5 CV files with fixed names:

    CV_1.pdf, CV_2.pdf, CV_3.pdf, CV_4.pdf, CV_5.pdf

Your system must process these CVs and output a list of 5 scores,
one score per CV, in the same order:

    scores = [s1, s2, s3, s4, s5]

Each score must be a float in the range [0, 1], representing the
reliability or confidence that the CV is valid (or meets the task criteria).

The ground-truth labels are binary:

    groundtruth = [0 or 1, ..., 0 or 1]

Each CV is evaluated independently using a threshold of 0.5:

- If score > 0.5 and groundtruth == 1 → Full credit
- If score ≤ 0.5 and groundtruth == 0 → Full credit
- Otherwise → No credit

In other words, 0.5 is the decision threshold.

- Each CV contributes equally.
- Final score = (number of correct decisions) / 5


```python
# =====================================================
#  Evaluation code
# =====================================================

def evaluate(scores, groundtruth, threshold=0.5):
    """
    scores: list of floats in [0, 1], length = 5
    groundtruth: list of ints (0 or 1), length = 5
    """
    assert len(scores) == 5
    assert len(groundtruth) == 5

    correct = 0
    decisions = []

    for s, gt in zip(scores, groundtruth):
        pred = 1 if s > threshold else 0
        decisions.append(pred)
        if pred == gt:
            correct += 1

    final_score = correct / len(scores)

    return {
        "decisions": decisions,
        "correct": correct,
        "total": len(scores),
        "final_score": final_score
    }
```

```python
import re
from typing import Any, Dict, Optional

def _maybe_int(x: Any) -> Optional[int]:
    if isinstance(x, int) and not isinstance(x, bool):
        return x
    if isinstance(x, str) and x.strip().isdigit():
        return int(x.strip())
    return None

def normalize_tool_args(tool_name: str, args: Dict[str, Any]) -> Dict[str, Any]:
    # 强制 FB 搜索 limit=5
    if tool_name == "search_facebook_users":
        out = dict(args)
        out["limit"] = 5
        return out

    # Facebook profile：只接受 int user_id
    if tool_name == "get_facebook_profile":
        uid = args.get("user_id", args.get("id"))
        uid_int = _maybe_int(uid)
        if uid_int is None:
            raise ValueError(f"get_facebook_profile requires integer user_id, got {uid!r}")
        return {"user_id": uid_int}

    # LinkedIn profile：只接受 int person_id
    if tool_name == "get_linkedin_profile":
        pid = args.get("person_id", args.get("id"))
        pid_int = _maybe_int(pid)
        if pid_int is None:
            raise ValueError(f"get_linkedin_profile requires integer person_id, got {pid!r}")
        return {"person_id": pid_int}

    return args
```

```python
def normalize_raw_candidates(raw):
    """
    Normalize tool-return raw structure into a list[dict] of candidates.
    Supports:
    - already list[dict] candidates
    - list[{"type":"text","text":"[...json...]"}]
    - {"type":"text","text":"[...json...]"}
    - string "[...json...]"
    - otherwise returns []
    """
    if raw is None:
        return []

    # Case 1: already list of dict candidates (id/display_name/...)
    if isinstance(raw, list):
        # list of blocks?
        if raw and isinstance(raw[0], dict) and "text" in raw[0] and isinstance(raw[0]["text"], str):
            try:
                parsed = json.loads(raw[0]["text"])
                return parsed if isinstance(parsed, list) else []
            except Exception:
                return []
        # list of candidates
        if raw and isinstance(raw[0], dict) and ("id" in raw[0] or "display_name" in raw[0] or "name" in raw[0]):
            return raw
        return []

    # Case 2: dict block
    if isinstance(raw, dict):
        if "text" in raw and isinstance(raw["text"], str):
            try:
                parsed = json.loads(raw["text"])
                return parsed if isinstance(parsed, list) else []
            except Exception:
                return []
        return []

    # Case 3: raw is a JSON string
    if isinstance(raw, str):
        try:
            parsed = json.loads(raw)
            return parsed if isinstance(parsed, list) else []
        except Exception:
            return []

    return []
```

```python
import json
from typing import Any, Dict, List, Tuple, Optional

from langchain_core.messages import SystemMessage, HumanMessage, ToolMessage, FunctionMessage
# Closed loop for chain 1 Chain 1 Need to retrieve information from cv file, using the seaerch function for both two media platforms
def _dump(x: Any) -> str:
    try:
        return json.dumps(x, ensure_ascii=False)
    except Exception:
        return str(x)

def _parse_json_maybe(s: Any) -> Dict[str, Any]:
    if isinstance(s, dict):
        return s
    if isinstance(s, str):
        try:
            return json.loads(s)
        except Exception:
            return {}
    return {}

async def run_prompt_loop_compat(
    *,
    llm_with_tools,
    tool_map: Dict[str, Any],
    system_text: str,
    user_text: str,
    user_vars: Dict[str, Any],
    max_rounds: int = 8,
) -> Tuple[str, List[Dict[str, Any]]]:

    messages: List[Any] = [
        SystemMessage(content=system_text),
        HumanMessage(content=user_text.format(**user_vars)),
    ]
    history: List[Dict[str, Any]] = []

    for _ in range(max_rounds):
        ai = await llm_with_tools.ainvoke(messages)
        messages.append(ai)

        # -------- 1) NEW: tool_calls format --------
        tool_calls = getattr(ai, "tool_calls", None) or []
        if tool_calls:
            for tc in tool_calls:
                tool_name = tc["name"]
                args = tc["args"]
                call_id = tc["id"]  # must exist in tool_calls mode
                try:
                    tool_res = await tool_map[tool_name].ainvoke(args)
                    history.append({"tool": tool_name, "args": args, "result": tool_res})
                    messages.append(ToolMessage(content=_dump(tool_res), tool_call_id=call_id))
                except Exception as e:
                    history.append({"tool": tool_name, "args": args, "error": str(e)})
                    messages.append(ToolMessage(content=_dump({"error": str(e)}), tool_call_id=call_id))
            continue  # go next round

        # -------- 2) OLD: function_call format --------

        # -------- 3) No tool call => final answer --------
        final = ai.content
        if isinstance(final, list):
            parts = []
            for it in final:
                if isinstance(it, dict) and "text" in it:
                    parts.append(it["text"])
                elif isinstance(it, str):
                    parts.append(it)
            final = "\n".join(parts)
        return str(final).strip(), history
    print(history)
    return "", history
```

```python
import json
import re
from typing import Any, Dict, List, Optional

from langchain_core.prompts import ChatPromptTemplate
from langchain_core.tools import ToolException

# ---------- robust content/json ----------
def _content(msg: Any) -> str:
    if hasattr(msg, "content"):
        c = msg.content
        if isinstance(c, str):
            return c.strip()
        if isinstance(c, list):
            parts = []
            for item in c:
                if isinstance(item, str):
                    parts.append(item)
                elif isinstance(item, dict) and "text" in item and isinstance(item["text"], str):
                    parts.append(item["text"])
            return "\n".join(parts).strip()
        return str(c).strip()
    return str(msg).strip()

def _json_load(s: str) -> Dict[str, Any]:
    s = (s or "").strip()
    if not s:
        return {}
    try:
        return json.loads(s)
    except Exception:
        m = re.search(r"\{.*\}", s, flags=re.S)
        return json.loads(m.group(0)) if m else {}

def _float01(x: Any) -> float:
    t = str(x).strip()
    m = re.search(r"(?<!\d)(0(?:\.\d+)?|1(?:\.0+)?)(?!\d)", t)
    if m:
        return float(m.group(1))
    m2 = re.search(r"([+-]?\d+(?:\.\d+)?)", t)
    if not m2:
        return 0.5
    v = float(m2.group(1))
    return max(0.0, min(1.0, v))
def label_score(label: str, mapping: dict, default=0.55) -> float:
    if not label:
        return default
    return mapping.get(label.lower().strip(), default)

def compute_score_v2(report: dict) -> float:
    # weights sum to 1.0
    weights = {
        "name": 0.30,
        "city": 0.10,
        "country": 0.30,
        "job": 0.15,
        "education": 0.15,
    }

    name_map = {"exact":1.0, "close":0.85, "unknown":0.55, "mismatch":0.0}
    geo_map  = {"same":1.0, "close":0.8, "unknown":0.55, "mismatch":0.0}
    job_map  = {"match":1.0, "strong":0.9, "partial":0.7, "unknown":0.55, "mismatch":0.0}
    edu_map  = {"match":1.0, "partial":0.7, "unknown":0.55, "mismatch":0.0}

    name_s = (report.get("name", {}) or {}).get("status", "unknown")
    city_s = (report.get("city", {}) or {}).get("status", "unknown")
    country_s = (report.get("country", {}) or {}).get("status", "unknown")
    job_s = (report.get("job", {}) or {}).get("status", "unknown")
    edu_s = (report.get("education", {}) or {}).get("status", "unknown")

    base = 0.0
    base += weights["name"] * label_score(name_s, name_map)
    base += weights["city"] * label_score(city_s, geo_map)
    base += weights["country"] * label_score(country_s, geo_map)
    base += weights["job"] * label_score(job_s, job_map)
    base += weights["education"] * label_score(edu_s, edu_map)

    # Hard floor: name exact + (city same/close OR country same/close) => >=0.50
    if name_s == "exact" and (city_s in ("same","close") or country_s in ("same","close")):
        base = max(base, 0.50)

    # Contradiction penalty (soft)
    contradictions = report.get("contradictions") or []
    n = len(contradictions)
    if n >= 3:
        base *= 0.65
    elif n == 2:
        base *= 0.80
    elif n == 1:
        base *= 0.90

    return max(0.0, min(1.0, base))
import re

def fix_name_location_glue(text: str) -> str:
    """
    Heuristic:
    - If the very first line contains letters immediately followed by a known 'City, Country' pattern,
      insert a newline before the city.
    - Also normalize 'Hometown' line if present.
    """
    lines = [ln.rstrip() for ln in (text or "").splitlines()]
    if not lines:
        return text

    first = lines[0].strip()

    # pattern: "...<Name><City>, <Country>" without space between name and city
    m = re.search(r"([A-Za-z]+)([A-Z][a-z]+,\s*[A-Z][a-z]+)", first)
    if m:
        # insert newline between the two groups
        fixed_first = first.replace(m.group(1) + m.group(2), m.group(1) + "\n" + m.group(2))
        # if replacement created a newline, split into multiple lines
        new_first_lines = fixed_first.split("\n")
        lines = new_first_lines + lines[1:]

    # normalize hometown line variants
    lines = [re.sub(r"^(.*)\s*\(Hometown\)\s*$", r"Hometown: \1", ln) for ln in lines]

    return "\n".join(lines)
#---- strong city score
def label_score(label: str, mapping: dict, default=0.55) -> float:
    if not label:
        return default
    return mapping.get(label.lower().strip(), default)

def compute_score_strong_identity(report: dict) -> float:
    # weights (identity heavy)
    weights = {
        "name": 0.35,
        "city": 0.20,
        "country": 0.15,   # city+country=0.35, 加上name=0.70
        "job": 0.20,
        "education": 0.10,
    }

    name_map = {"exact":1.0, "close":0.85, "unknown":0.55, "mismatch":0.0}
    geo_map  = {"same_city":1.0, "same":1.0, "close_city":0.85, "close":0.8, "same_country":0.75, "unknown":0.55, "mismatch":0.0}
    job_map  = {"match":1.0, "strong":0.9, "partial":0.7, "unknown":0.55, "mismatch":0.0}
    edu_map  = {"match":1.0, "partial":0.7, "unknown":0.55, "mismatch":0.0}

    name_s = (report.get("name", {}) or {}).get("status", "unknown")
    city_s = (report.get("city", {}) or {}).get("status", "unknown")
    country_s = (report.get("country", {}) or {}).get("status", "unknown")
    job_s = (report.get("job", report.get("experience", {})) or {}).get("status", "unknown")
    edu_s = (report.get("education", {}) or {}).get("status", "unknown")

    base = 0.0
    base += weights["name"] * label_score(name_s, name_map)
    base += weights["city"] * label_score(city_s, geo_map)
    base += weights["country"] * label_score(country_s, geo_map)
    base += weights["job"] * label_score(job_s, job_map)
    base += weights["education"] * label_score(edu_s, edu_map)

    # ✅ hard floor: name exact AND (city matches OR country matches) => at least 0.50
    geo_ok = (city_s in ("same_city","same","close_city","close")) or (country_s in ("same_country","same","close"))
    if name_s == "exact" and geo_ok:
        base = max(base, 0.50)

    # contradiction penalty: soften it (do NOT destroy identity match)
    contradictions = report.get("contradictions") or []
    n = len(contradictions)
    if n >= 3:
        base *= 0.80
    elif n == 2:
        base *= 0.88
    elif n == 1:
        base *= 0.94

    # optional cap: if job+education both mismatch AND contradictions>=2, cap to avoid false positives
    if job_s == "mismatch" and edu_s == "mismatch" and n >= 2:
        base = min(base, 0.58)

    return max(0.0, min(1.0, base))

# ---------- tool guard ----------
tool_map = {t.name: t for t in tools}

def _maybe_int(x: Any) -> Optional[int]:
    if isinstance(x, int) and not isinstance(x, bool):
        return x
    if isinstance(x, str) and x.strip().isdigit():
        return int(x.strip())
    return None

def normalize_tool_args(tool_name: str, args: Dict[str, Any]) -> Dict[str, Any]:
    if tool_name == "search_facebook_users":
        out = dict(args); out["limit"] = 5
        return out
    if tool_name == "get_facebook_profile":
        uid_int = _maybe_int(args.get("user_id"))
        if uid_int is None:
            raise ValueError(f"get_facebook_profile requires int user_id, got {args.get('user_id')!r}")
        return {"user_id": uid_int}
    if tool_name == "get_linkedin_profile":
        pid_int = _maybe_int(args.get("person_id"))
        if pid_int is None:
            raise ValueError(f"get_linkedin_profile requires int person_id, got {args.get('person_id')!r}")
        return {"person_id": pid_int}
    return args

async def tool_invoke(tool_name: str, args: Dict[str, Any]) -> Any:
    if tool_name not in tool_map:
        raise ValueError(f"Unknown tool: {tool_name}")
    safe_args = normalize_tool_args(tool_name, args)
    return await tool_map[tool_name].ainvoke(safe_args)

def _tool_calls(msg: Any) -> List[Dict[str, Any]]:
    return getattr(msg, "tool_calls", []) or []
```
